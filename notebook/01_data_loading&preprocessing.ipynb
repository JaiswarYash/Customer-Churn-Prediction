### Project Name:- Customer Churn Prediction 

### Problem :-
Retaining existing customers is significantly more cost-effective than acquiring new ones, yet many businesses lack the ability to identify "at-risk" customers before they leave. Undetected churn gradually damages brand trust, reduces lifetime customer value, and weakens long-term growth.  

### Solution:-
This project Aim to solve that problem by building a prediction system that help to detect customers who likely to churn -  early enough for the business to act."

### 02. Data preprocessing

```python
print(dataset_1.info())
```

```python
print(dataset_2.info()) 
```

```python
# changing the total charges column to float64
dataset_2['TotalCharges'] = pd.to_numeric(dataset_2['TotalCharges'], errors='coerce')
print(dataset_2.info())
```

```python
# step 2: merging the datasets
merged_dataset = pd.concat([dataset_1, dataset_2], axis=0)
print(merged_dataset.shape)
merged_dataset.head(10)
```

```python
print(merged_dataset.info() ) 
```

```python
# checking for missing values
print(merged_dataset.isnull().sum())
```

```python
# checking for Null valus
print(merged_dataset.isnull().sum())
```

```python
# 1. mapping churn column to binary values
merged_dataset['Churn'] = merged_dataset['Churn'].map({'Yes': 1, 'No': 0})

# 2. mapping gender column to binary values
merged_dataset['gender'] = merged_dataset['gender'].apply(lambda x: 1 if x == 'Male' else 0)


merged_dataset.head(10)
```

```python
# replacing No internet service or No phone service with No
cols = ['MultipleLines', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies']

for col in cols:
    merged_dataset[col] = merged_dataset[col].replace({'No internet service': 'No', 'No phone service': 'No'})

merged_dataset.head(10)


#reason:
#1. No internet service and No phone service are essentially the same as No for the respective services.
#2. This replacement simplifies the dataset and reduces the number of unique values in these columns, 
# which can help improve the performance of machine learning models by reducing complexity and potential noise in the data.
# 3. It also makes it easier to interpret the data, as it clearly indicates that the customer does not have the service, 
# rather than having a separate category for those who do not have internet or phone service.

 
```

```python
# maping the categorical columns to numerical values
cols = ['Partner', 'Dependents', 'PhoneService', 'PaperlessBilling', 'MultipleLines', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies']

for col in cols:
    merged_dataset[col] = merged_dataset[col].map({'Yes': 1, 'No': 0})

merged_dataset.head(10)
```

```python
# droping the ustomerId and TotalCharges columns
merged_dataset.drop(['customerID', 'TotalCharges'], axis=1, inplace=True)
print(merged_dataset.info())
```
