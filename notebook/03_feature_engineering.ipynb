## 03 Feature Engineering

```python
import numpy as np
import pandas as pd
import os
```

```python
# Load the dataset
df = pd.read_csv('../data/processed_data/clean_data.csv')

# check info
df.info()
```

```python
df.head(10)
```

```python
# encoding categorical variables

df_encoded = df.copy()
```

```python
#one-hot encoding
df_encoded = pd.get_dummies(df_encoded, columns=['InternetService', 'PaymentMethod'], drop_first=True)

# ordinal encoding
df_encoded['Contract'] = df_encoded['Contract'].map({'Month-to-month': 0, 'One year': 1, 'Two year': 2})

```

```python
cols = ["InternetService_Fiber optic", "InternetService_No",
        "PaymentMethod_Bank transfer (automatic)", 
        "PaymentMethod_Credit card",
        "PaymentMethod_Credit card (automatic)",
        "PaymentMethod_Electronic check", 
        "PaymentMethod_Mailed check"]

df_encoded[cols] = df_encoded[cols].astype(int)
```

```python
df_encoded.info()
```

```python
df_encoded.head(10)
```

```python
corr = df_encoded.corr(numeric_only=True)
print(corr['Churn'].sort_values(ascending=False))
```

```python
# droping columns with low correlation
df_encoded.drop(columns=['gender','PaymentMethod_Mailed check','OnlineBackup', 'DeviceProtection','PhoneService','MultipleLines','StreamingMovies','PaymentMethod_Credit card', 'StreamingTV'], inplace=True)
```

```python
df_encoded.info()
```

```python
df_encoded.head(10)
```

```python
# skewness
numerical_cols = ['tenure', 'MonthlyCharges']
print(df_encoded[numerical_cols].skew())
```

Both values are between -0.5 and 0.5 — that's the normal range, no transformation needed at all.

```python
# creating directory if it doesn't exist
if not os.path.exists('DATA/processed'):
    os.makedirs('DATA/processed')

# save full clean dataset 
df_encoded.to_csv('../data/processed_data/featured_data.csv', index=False)
```
